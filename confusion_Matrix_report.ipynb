```python
import numpy as np
from sklearn.metrics import confusion_matrix,classification_report
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
actual    = np.array(
  ['Dog','Dog','Dog','Not Dog','Dog','Not Dog','Dog','Dog','Not Dog','Not Dog'])
predicted = np.array(
  ['Dog','Not Dog','Dog','Not Dog','Dog','Dog','Dog','Dog','Not Dog','Not Dog'])
```

```python
print(len(actual))
print(len(predicted))
```

```python
cm = confusion_matrix(actual,predicted)
print(cm)
# 5 - actual dog - prediction dog 
# 3 - not a dog in actual - not a dog in prediction 
# 1 - actual it is a dog - not a dog in prediction 
# 1 - actual it is not a dog - it is a dog in prediction 
```

```python
sns.heatmap(cm, 
            annot=True,
            fmt='g',  # fmt - ?
            xticklabels=['Dog','Not Dog'],
            yticklabels=['Dog','Not Dog'])
plt.ylabel('Actual', fontsize=13)
plt.title('Confusion Matrix', fontsize=17, pad=20)
plt.gca().xaxis.set_label_position('top')  # output - prediction ka position 
plt.xlabel('Prediction', fontsize=13)
plt.gca().xaxis.tick_top() # Dog & Not Dog - position 

# plt.gca().figure.subplots_adjust(bottom=0.1)
# plt.gca().figure.text(0.5, 0.05, 'Prediction', ha='center', fontsize=13) # output - prediction 
plt.show()
```

```python
print(classification_report(actual, predicted))  
```
